```python
import pandas as pd

df_auto = pd.read_csv('Automobile.csv')
display(df_auto.head())
```

```python
# gráfico de dispersión
plt.figure(figsize=(8, 5))
plt.scatter(df_auto['weight'], df_auto['mpg'], color='deeppink', alpha=0.6, edgecolors='pink')

plt.title('Relación entre Peso (weight) y Consumo (mpg)')
plt.xlabel('Peso (weight en lbs)')
plt.ylabel('Millas por Galón (mpg)')
plt.grid(True, linestyle='--', alpha=0.3)

plt.show()
```

Muestra una clara relación negativa, a mayor peso del vehículo, menor es su rendimiento de combustible (menos millas por galón)


```python
# histograma de mpg
plt.figure(figsize=(8, 5))
plt.hist(df_auto['mpg'], bins=15, color='deeppink', edgecolor='white', alpha=0.7)

plt.title('Distribución del Consumo de Combustible (mpg)')
plt.xlabel('Millas por Galón (mpg)')
plt.ylabel('Frecuencia (Cantidad de vehículos)')
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Nos aseguramos de mapear la columna a valores numéricos simples
if df_auto['origin'].dtype == 'object':
    # Si viene como texto, asignamos un número único a cada país
    origen_valores = df_auto['origin'].map({'usa': 1, 'japan': 2, 'europe': 3,
                                            'USA': 1, 'Japan': 2, 'Europe': 3})
else:
    origen_valores = df_auto['origin']

# Generar UNA SOLA caja para la variable origin
plt.figure(figsize=(6, 5))
caja = plt.boxplot(origen_valores.dropna(), patch_artist=True)

# Personalizar en estilo rosa
for box in caja['boxes']:
    box.set_facecolor('pink')
    box.set_color('deeppink')
    box.set_linewidth(1.5)

for median in caja['medians']:
    median.set_color('crimson')
    median.set_linewidth(2)

plt.title('Diagrama de Caja: Origen (origin)')
plt.xticks([1], ['Origen'])
plt.ylabel('Código de Origen (1: USA, 2: Europa, 3: Asia)')
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.show()
```

Al ubicarse la caja y la mediana muy apretadas en el valor 1, el gráfico demuestra estadísticamente que el dataset está dominado por vehículos de origen norteamericano, con una presencia mucho menor de modelos europeos y asiáticos.

```python
# gráfico de barras
conteo = df_auto.groupby(['model_year', 'origin']).size().unstack(fill_value=0)

plt.figure(figsize=(12, 6))
conteo.plot(kind='bar', figsize=(12, 6), color=['deeppink', 'hotpink', 'pink'], edgecolor='black')

plt.title('Cantidad de Vehículos por Año de Modelo y Origen')
plt.xlabel('Año del Modelo (model_year)')
plt.ylabel('Cantidad de Vehículos')
plt.legend(title='Origen')
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.show()
```

Estados Unidos (el rosa más clarito) lideró ampliamente el volumen de vehículos en casi todos los años analizados, mientras que Europa y Asia  mantienen volúmenes de producción bajos pero bastnte constantes a lo largo del tiempo.
